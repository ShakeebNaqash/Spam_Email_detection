#  Spam Email Detection using TF-IDF
This notebook detects spam and ham emails using TF-IDF vectorization and a Naive Bayes classifier.

```python

# Install necessary libraries # !pip install pandas scikit-learn streamlit joblib
```

```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix
```

```python

# Read the CSV file directly (ensure it's in the same directory)
df = pd.read_csv('190KDataset.csv')

# Display first few rows
df.head()
```

```python

# Clean and prepare the data
print("Columns:", df.columns)
df.columns = ['label', 'text']  # Adjust column names if needed
df['label'] = df['label'].map({'ham': 0, 'spam': 1})
df.dropna(inplace=True)
```

```python
df = pd.read_csv("190KDataset.csv")
print(df.head())
print(df.columns)
```

```python
print(df['label'].unique())

df['label'] = df['label'].astype(str).str.strip().str.lower()
print(df['label'].unique())

df['label'] = df['label'].map({'ham': 0, 'spam': 1})
df.dropna(subset=['label', 'text'], inplace=True)
print(df.shape)
```

```python

# Split and vectorize using TF-IDF
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.2, random_state=42)

tfidf = TfidfVectorizer(stop_words='english')
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
```

```python

# Train the Naive Bayes classifier
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)
```

```python

# Evaluate the model
y_pred = model.predict(X_test_tfidf)
print("Classification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python

# Test the model on custom inputs
custom_emails = [
    "Congratulations! You've won a $1000 Walmart gift card. Click here to claim now.",
    "Hi John, please review the report and send your feedback before tomorrow's meeting.",
    "You are selected for a lottery of 50,000 dollars. Provide your bank details now!"
]

custom_tfidf = tfidf.transform(custom_emails)
custom_preds = model.predict(custom_tfidf)

for email, pred in zip(custom_emails, custom_preds):
    label = "Spam" if pred == 1 else "Ham"
    print(f"\nEmail: {email}\nPrediction: {label}")
```

```python
import joblib

# Save model
joblib.dump(model, "model/spam_classifier.pkl")

# Save TF-IDF vectorizer
joblib.dump(tfidf, "model/tfidf_vectorizer.pkl")
```
